```python
#soal 3.1
def hitung_luas_persegi(sisi):
  hasil = sisi * sisi
  print(f"Luas persegi adalah: {hasil}")
  return hasil

def rata_rata(nilai1, nilai2, nilai3):
  rata_rata = (nilai1 + nilai2 + nilai3) / 3
  return rata_rata

nilai1 = hitung_luas_persegi(5)
nilai2 = hitung_luas_persegi(10)
nilai3 = hitung_luas_persegi(15)

hasil_rata = rata_rata(nilai1, nilai2, nilai3)
print(f"Rata-rata 3 persegi adalah: {hasil_rata}")
```

```python
#soal 5
def konversi_nilai(nilai_angka):
  if nilai_angka >=86:
    return "A"
  elif nilai_angka >=76:
    return "AB"
  elif nilai_angka >=66:
    return "B"
  else:
    return "C"
nilai_input = int(input("Masukkan nilai: "))
indeks = konversi_nilai(nilai_input)
print(f"Nilai huruf anda adalah: {indeks}")
```
